### Step1 : Import libraries and setup api call

```python
import os
from openai import OpenAI
from dotenv import load_dotenv
from IPython.display import display, Markdown
```

#### We want the OS to grab the API Key and give it to the OpenAI module

```python
#You first need to load in the env variables: in .env is the key
load_dotenv()

#I am saying that os (Mac) needs to use the function getenv 
#I pass the variable name "OpenAI_API_KEY" for the function to find from .env
#then I assign the value to the variable OPENAI_API_KEY
OpenAI_API_KEY = os.getenv("OPENAI_API_KEY")

if OpenAI_API_KEY is None:
    raise Exception("API Key is Missing.")
else:
    print(OpenAI_API_KEY[:8])
```

```python
#establishing OpenAI object from OpenAI module
#Open AI module has the class OpenAI and client is an object of that class
client = OpenAI ()

response_summary = client.chat.completions.create(
    model = "gpt-4o-search-preview",
    messages= [
        {"role": "user", "content" :"What is today's date? Tell me the top 3 AI news in the past week. Make sure you are choosing reliable popular News sources. Make sure to include dates."}
    ],
    max_tokens= 500
)


reply_summary = response_summary.choices[0].message.content

display(Markdown(reply_summary))
```

#### Summarizing for the Linkedin Post

```python
system_message = """ You are a helpful assistant. You create LinkedIn Posts based on the text provided to you.
                  Make sure to create the post so it shows that I am educated AI engineer who is very up to date with the current times.
 """

prompt = f""" Take the following new sources and write me a LinkedIn post.
{reply_summary}
"""

response_post = client.chat.completions.create(
    model= "gpt-4o-mini",
    messages= [
        {"role" : "system", "content": system_message},
        {"role": "user", "content" : prompt}
    ]
)

reply_post = response_post.choices[0].message.content
display(Markdown(reply_post))
```
